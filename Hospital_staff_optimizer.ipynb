```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
print("Hospital needs to determine the required:")
print("1. Doctors")
print("2. Nurses")
print("3. Support Staff")
print()

print("We use a system of linear equations.")
```

```python
A = np.array([
    [1, 1, 1],
    [2, 1, 3],
    [1, 3, 2]
], dtype=float)

B = np.array([20, 40, 35], dtype=float)

print("Equations:")
print("x + y + z = 20")
print("2x + y + 3z = 40")
print("x + 3y + 2z = 35")
```

```python
def gauss_elimination(A, B):
    n = len(B)

    aug = np.column_stack((A, B))

    for i in range(n):

        max_row = i + np.argmax(abs(aug[i:, i]))

        if max_row != i:
            aug[[i, max_row]] = aug[[max_row, i]]

        for j in range(i + 1, n):
            factor = aug[j, i] / aug[i, i]
            aug[j] = aug[j] - factor * aug[i]

    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        x[i] = (
            aug[i, -1] -
            np.dot(aug[i, i + 1:n], x[i + 1:n])
        ) / aug[i, i]

    return x
```

```python
solution = gauss_elimination(A, B)

doctors = solution[0]
nurses = solution[1]
support_staff = solution[2]

print("Solution:")
print("Doctors =", doctors)
print("Nurses =", nurses)
print("Support Staff =", support_staff)
```

```python
print("=" * 50)
print("       FINAL STAFFING REQUIREMENT")
print("=" * 50)

print(f"Doctors       : {doctors:.2f}")
print(f"Nurses        : {nurses:.2f}")
print(f"Support Staff : {support_staff:.2f}")
```

```python
eq1 = doctors + nurses + support_staff
eq2 = 2*doctors + nurses + 3*support_staff
eq3 = doctors + 3*nurses + 2*support_staff

print("Verification:")
print(f"Equation 1 = {eq1:.2f}")
print(f"Equation 2 = {eq2:.2f}")
print(f"Equation 3 = {eq3:.2f}")
```

```python
staff = ["Doctors", "Nurses", "Support Staff"]
values = [doctors, nurses, support_staff]

plt.figure(figsize=(8,5))

bars = plt.bar(staff, values)

plt.title("Hospital Ward Staffing Requirements")
plt.xlabel("Staff Category")
plt.ylabel("Number of Staff")

for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        value,
        f"{value:.2f}",
        ha="center",
        va="bottom"
    )

plt.show()
```
